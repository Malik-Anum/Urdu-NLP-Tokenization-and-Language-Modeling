**Step 4: Unigram Language Model (Without Smoothing)**

```python
## Unigram Model Without Smoothing

class UnigramModel:

    def __init__(self, sentences):
        self.word_counts = Counter()
        self.total_words = 0

        for sentence in sentences:
            self.word_counts.update(sentence)
            self.total_words += len(sentence)

    def getProbability(self, word):
        if word in self.word_counts:
            return self.word_counts[word] / self.total_words
        else:
            return 0

    def getSentenceProbability(self, sentence):
        probability = 1

        for word in sentence:
            probability *= self.getProbability(word)

        return probability

    def generateSentence(self, length=10):
        words = list(self.word_counts.keys())
        return " ".join(random.choice(words) for _ in range(length))
```

**Step 7: Smoothed Unigram Model (Laplace Smoothing)**

```python
## Smoothed Unigram Model

class SmoothedUnigramModel:

    def __init__(self, sentences):
        self.word_counts = Counter()
        self.total_words = 0

        for sentence in sentences:
            self.word_counts.update(sentence)
            self.total_words += len(sentence)

        self.vocabulary = set(self.word_counts.keys())
        self.vocab_size = len(self.vocabulary)

    def getProbability(self, word):
        count = self.word_counts[word]
        return (count + 1) / (self.total_words + self.vocab_size)

    def getSentenceProbability(self, sentence):
        probability = 1

        for word in sentence:
            probability *= self.getProbability(word)

        return probability

    def generateSentence(self, length=10):
        words = list(self.vocabulary)
        return " ".join(random.choice(words) for _ in range(length))
```

**Step 8: Bigram Language Model (Without Smoothing)**

```python
## Bigram Model Without Smoothing

class BigramModel:

    def __init__(self, sentences):
        self.bigram_counts = Counter()
        self.unigram_counts = Counter()

        for sentence in sentences:
            self.unigram_counts.update(sentence)

            for i in range(len(sentence)-1):
                bigram = (sentence[i], sentence[i+1])
                self.bigram_counts[bigram] += 1


    def getProbability(self, word1, word2):

        bigram = (word1, word2)

        if self.unigram_counts[word1] == 0:
            return 0

        return self.bigram_counts[bigram] / self.unigram_counts[word1]


    def getSentenceProbability(self, sentence):

        probability = 1

        for i in range(len(sentence)-1):
            probability *= self.getProbability(
                sentence[i],
                sentence[i+1]
            )

        return probability


    def generateSentence(self):

        sentence = ["<s>"]
        current_word = "<s>"

        while True:

            possible_words = []

            for (word1, word2) in self.bigram_counts.keys():
                if word1 == current_word:
                    possible_words.append(word2)

            if len(possible_words) == 0:
                break

            next_word = random.choice(possible_words)

            if next_word == "</s>":
                break

            sentence.append(next_word)
            current_word = next_word

        return " ".join(sentence[1:])
```

**Step 10: Smoothed Bigram Model (Laplace Smoothing)**

```python
## Smoothed Bigram Model (Laplace Smoothing)

class SmoothedBigramModel:

    def __init__(self, sentences):

        self.bigram_counts = Counter()
        self.unigram_counts = Counter()
        self.vocabulary = set()

        for sentence in sentences:

            self.unigram_counts.update(sentence)
            self.vocabulary.update(sentence)

            for i in range(len(sentence)-1):
                bigram = (sentence[i], sentence[i+1])
                self.bigram_counts[bigram] += 1


    def getProbability(self, word1, word2):

        bigram = (word1, word2)

        bigram_count = self.bigram_counts[bigram]
        word_count = self.unigram_counts[word1]

        vocabulary_size = len(self.vocabulary)

        return (bigram_count + 1) / (word_count + vocabulary_size)


    def getSentenceProbability(self, sentence):

        probability = 1

        for i in range(len(sentence)-1):

            probability *= self.getProbability(
                sentence[i],
                sentence[i+1]
            )

        return probability


    def generateSentence(self):

        sentence = ["<s>"]
        current_word = "<s>"

        while True:

            possible_words = []

            for (word1, word2) in self.bigram_counts.keys():

                if word1 == current_word:
                    possible_words.append(word2)


            if len(possible_words) == 0:
                break


            next_word = random.choice(possible_words)


            if next_word == "</s>":
                break


            sentence.append(next_word)
            current_word = next_word


        return " ".join(sentence[1:])
```

**Step 12: Perplexity Calculation**

```python
def calculate_perplexity(model, sentences):
    total_words = 0
    total_log_probability = 0

    for sentence in sentences:
        probability = model.getSentenceProbability(sentence)

        if probability == 0:
            probability = 1e-10

        total_log_probability += math.log(probability)
        total_words += len(sentence)

    perplexity = math.exp(-total_log_probability / total_words)

    return perplexity
```

Q6: What happens when your unsmoothed bigram model encounters a bigram at test time that was never seen during training? How does the Laplace-smoothed bigram model handle this differently?

Asnwer: In an unsmoothed bigram model, if a bigram appears in the test corpus but was never observed during training, its probability becomes zero.

The probability is calculated as:

$$ P(w_2|w_1)=\frac{Count(w_1,w_2)}{Count(w_1)} $$

If the bigram count is zero, the probability becomes:

$$ P(w_2|w_1)=0 $$

This causes the complete sentence probability to become zero, making the model unable to handle unseen word combinations.

For example, if the test corpus contains a bigram such as:
"computer galaxy"

and this bigram was not present in the training corpus, the unsmoothed bigram model assigns:

P(galaxy | computer) = 0

The Laplace-smoothed bigram model solves this problem by adding one to the bigram count:

$$ P(w_2|w_1)=\frac{Count(w_1,w_2)+1}{Count(w_1)+V} $$

where V is the vocabulary size.
Therefore, even unseen bigrams receive a small non-zero probability. This prevents sentence probabilities from becoming zero and allows the model to handle new word combinations more effectively.

**Step 4: Create Initial BPE Vocabulary**

```python
from docx import Document

# Load wordlist again
doc = Document("wordlist datset.docx")

wordlist = []

for para in doc.paragraphs:
    if para.text.strip():
        wordlist.append(para.text.strip())

print("Number of words:", len(wordlist))
print("First 10 words:")
print(wordlist[:10])
```

```python
from collections import Counter

# Create initial BPE vocabulary

bpe_vocab = {}

for word in wordlist:
    chars = list(word)
    bpe_vocab[" ".join(chars)] = 1


print("Initial BPE Vocabulary Size:", len(bpe_vocab))

print("\nFirst 10 vocabulary entries:")

for word in list(bpe_vocab.keys())[:10]:
    print(word)
```

**Step 5: Calculate Character Pair Frequencies**

```python
from collections import Counter

def get_pair_counts(vocab):
    pairs = Counter()

    for word in vocab:
        symbols = word.split()

        for i in range(len(symbols)-1):
            pair = (symbols[i], symbols[i+1])
            pairs[pair] += 1

    return pairs


pair_counts = get_pair_counts(bpe_vocab)

print("Top 10 frequent character pairs:\n")

for pair, count in pair_counts.most_common(10):
    print(pair, ":", count)
```

**Step 6: Perform BPE Merge Operations**

```python
# Step 6: BPE Merge Operations

def merge_pair(pair, vocab):
    new_vocab = {}

    bigram = " ".join(pair)
    replacement = "".join(pair)

    for word in vocab:
        new_word = word.replace(bigram, replacement)
        new_vocab[new_word] = vocab[word]

    return new_vocab


# Store learned merges
merges = []

# Perform 10 merges
for i in range(10):

    pair_counts = get_pair_counts(bpe_vocab)

    if not pair_counts:
        break

    best_pair = pair_counts.most_common(1)[0][0]

    merges.append(best_pair)

    print(f"Merge {i+1}: {best_pair}")

    bpe_vocab = merge_pair(best_pair, bpe_vocab)
```

**Step 7: Apply Learned BPE Merges for Segmentation**

```python
# Step 7: Apply BPE segmentation

def apply_bpe(word, merges):
    symbols = list(word)

    for merge in merges:
        new_symbols = []
        i = 0

        while i < len(symbols):
            if i < len(symbols)-1 and (symbols[i], symbols[i+1]) == merge:
                new_symbols.append(symbols[i] + symbols[i+1])
                i += 2
            else:
                new_symbols.append(symbols[i])
                i += 1

        symbols = new_symbols

    return " ".join(symbols)


segmented_sentences = []

for sentence in sentences:
    segmented_sentences.append(apply_bpe(sentence, merges))


print("First 5 segmented sentences:\n")

for i, s in enumerate(segmented_sentences[:5]):
    print(i+1, s)
```

**Step 8: Save Segmented Output File**

```python
# Step 8: Save segmented output

with open("segmented_output.txt", "w", encoding="utf-8") as f:
    for sentence in segmented_sentences:
        f.write(sentence + "\n")

print("segmented_output.txt saved successfully")
```

##Task P3T3 — Urdu Text Tokenization and N-Gram Generation

Step 1: Load Urdu Corpus

```python
from docx import Document

# Load Urdu corpus from P3T2

doc = Document("urdu_corpus.docx")

text = ""

for para in doc.paragraphs:
    text += para.text + " "

print("Total characters:", len(text))

print("\nSample text:")
print(text[:500])
```

Step 2: Urdu Tokenization and Preprocessing

```python
import re
import unicodedata

# Step 2: Urdu text preprocessing

def normalize_urdu(text):
    # Unicode normalization
    text = unicodedata.normalize("NFKC", text)

    # Normalize different forms of ک
    text = text.replace("ك", "ک")

    # Normalize different forms of ی
    text = text.replace("ي", "ی")

    return text


def clean_urdu_text(text):

    # Normalize unicode
    text = normalize_urdu(text)

    # Remove English letters, numbers and special symbols
    text = re.sub(r"[A-Za-z0-9]", "", text)

    # Keep Urdu characters, spaces only
    text = re.sub(r"[^\u0600-\u06FF\s]", " ", text)

    # Remove extra spaces
    text = re.sub(r"\s+", " ", text).strip()

    return text


# Apply cleaning
clean_text = clean_urdu_text(text)

# Tokenization (whitespace based)
tokens = clean_text.split()


print("Total tokens after preprocessing:", len(tokens))

print("\nFirst 20 tokens:")
print(tokens[:20])
```

Step 3: Unigram Generation (Top 20)

```python
from collections import Counter

# Step 3: Generate Unigrams

unigram_counts = Counter(tokens)

top_20_unigrams = unigram_counts.most_common(20)

print("Top 20 Unigrams:\n")

for word, frequency in top_20_unigrams:
    print(word, ":", frequency)
```

Step 4: Bigram Generation (Top 20)

```python
from collections import Counter

# Step 4: Generate Bigrams

bigrams = []

for i in range(len(tokens)-1):
    bigram = (tokens[i], tokens[i+1])
    bigrams.append(bigram)


bigram_counts = Counter(bigrams)

top_20_bigrams = bigram_counts.most_common(20)

print("Top 20 Bigrams:\n")

for pair, frequency in top_20_bigrams:
    print(pair, ":", frequency)
```

Step 5: Trigram Generation (Top 20)

```python
from collections import Counter

# Step 5: Generate Trigrams

trigrams = []

for i in range(len(tokens)-2):
    trigram = (tokens[i], tokens[i+1], tokens[i+2])
    trigrams.append(trigram)


trigram_counts = Counter(trigrams)

top_20_trigrams = trigram_counts.most_common(20)

print("Top 20 Trigrams:\n")

for tri, frequency in top_20_trigrams:
    print(tri, ":", frequency)
```
